```python
import pandas, warnings

warnings.filterwarnings('ignore')

sunspotsurl = '/content/sunspots.csv'
```

```python
sunspots = pandas.read_table(sunspotsurl, sep=',')

sunspots.head()
```

```python
sunspots.drop(sunspots.columns[0], axis = 1, inplace = True)

sunspots.head()
```

```python
sunspots['Date'].dtype
```

```python
sunspots['Date'] = pandas.to_datetime(sunspots['Date'])

sunspots['Date'].dtype
```

```python
sunspots.set_index('Date', inplace = True)

sunspots.index.name = 'Monthly'

sunspots.columns = ['Total']

sunspots.head()
```

```python
total_bulan = sunspots.shape[0]

total_bulan
```

```python
sunspots.tail()
```

```python
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = (15, 5)

plt.style.use('seaborn-v0_8-whitegrid')

ax = sunspots.plot.line()

ax.set_title("Total Kemunculan Sunspot Tahun 1769 - 2018", pad = 10)

ax.set_xlabel('Year', labelpad = 10)

ax.set_ylabel('Total Sunspot', labelpad = 10);
```
